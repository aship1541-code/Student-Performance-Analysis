```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("StudentsPerformance.csv")

df.head()
```

```python
print("--- Statistical Summary ---")
display(df.describe())
```

```python
print("Average Marks:")
print(df[['math score', 'reading score', 'writing score']].mean())

print("\nMath score average by Gender:")
print(df.groupby('gender')['math score'].mean())
```

```python
df['math score'].hist(bins=20, color='lightgreen', edgecolor='black')
plt.title('Distribution of Math Scores')
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.show()
```

```python
print("--- Average Scores by Parental Education ---")
parent_analysis = df.groupby('parental level of education')[['math score', 'reading score', 'writing score']].mean()
display(parent_analysis)

print("\n--- Impact of Test Preparation Course ---")
prep_analysis = df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()
display(prep_analysis)
```

Project Insights: Student Performance Analysis
1. Impact of Test Preparation Course
Students who completed the test preparation course performed significantly better across all three subjects compared to those who did not take any course.

Math Score: Completed (69.7) vs None (64.1)

Reading Score: Completed (73.9) vs None (66.5)

Writing Score: Completed (74.4) vs None (64.5)

Conclusion: The preparation course has a clear positive impact on student grades.

2. Parental Level of Education and Performance
There is a visible correlation between parental education and student scores.

Students whose parents have a Master's Degree or Bachelor's Degree tend to achieve the highest average scores in all areas.

The highest average reading (75.4) and writing (75.7) scores were recorded by students of parents with a Master's Degree.

Students whose parents only attended High School or Some High School recorded the lowest average scores.

3. Gender-Based Performance (From Visualization)
Based on the earlier analysis and graphs:

Female students generally outperform male students in Reading and Writing.

Male students show a slight advantage in Math scores.

4. Distribution of Math Scores
The histogram shows that the majority of students scored between 60 and 80 marks in Math. The distribution follows a "Normal Distribution" (Bell Curve), with very few students scoring below 20 or achieving a perfect 100.
